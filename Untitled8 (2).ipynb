Q1) What is a Support Vector Machine (SVM)?

Ans) A Support Vector Machine (SVM) is a supervised machine learning algorithm used for classification and regression tasks. It works by finding the optimal hyperplane that best separates different classes in the data.

How SVM Works (For Classification)

Separating Data: It tries to find the best boundary (hyperplane) that divides the data points into different categories.

Maximizing the Margin: SVM doesn’t just find any boundary—it finds the one that maximizes the distance (margin) between the closest data points from each class, called support vectors.

Handling Non-Linearly Separable Data: If the data isn't easily separable with a straight line, SVM uses a kernel trick (like polynomial or radial basis function (RBF) kernels) to map the data into a higher-dimensional space where it becomes linearly separable.

Why Use SVM?

Works well for both linear and non-linear classification problems.

Effective in high-dimensional spaces (e.g., text classification).

Robust to outliers because it focuses on support vectors rather than all data points.

Q2) What is the difference between Hard Margin and Soft Margin SVM ?


Ans) The difference between Hard Margin SVM and Soft Margin SVM lies in how strictly the algorithm separates data and handles outliers.

1. Hard Margin SVM

Strictly separates classes: It only works when the data is perfectly linearly separable (i.e., there are no overlapping classes).

Maximizes the margin: It finds the hyperplane that maximizes the margin between the closest points from each class (support vectors).

No tolerance for misclassification: Every data point must be correctly classified, making it sensitive to noise and outliers.

Limitation: If the data is not perfectly separable, Hard Margin SVM fails because no valid hyperplane can be found.

2. Soft Margin SVM

Allows misclassification: It introduces a slack variable (ξ) to permit some points to be on the wrong side of the margin or even misclassified.

Balances margin and misclassification: A parameter C controls this trade-off:
High C → Less tolerance for errors (acts more like Hard Margin).

Low C → More tolerance for errors (wider margin, better for noisy data).

Works with non-linearly separable data: It can still find a good decision boundary even if some points are wrongly classified.

3. Key Properties of Support Vectors

🔹 Only a subset of data points become support vectors → The model doesn’t rely on all data points, making SVM computationally efficient.
🔹 If support vectors change, the decision boundary changes → Removing or modifying a support vector can significantly impact the classifier.

4. Why Are Support Vectors Important?

✔ Generalization: SVMs maximize the margin using only the support vectors, leading to better generalization on unseen data.
✔ Robustness: Since only a few data points define the boundary, SVMs are less prone to overfitting, especially in high-dimensional spaces.

Q15) When should you use Gaussian Naïve Bayes over other variants?

Ans) When to Use Gaussian Naïve Bayes Over Other Variants

You should use Gaussian Naïve Bayes (GNB) when your features are continuous numerical values and approximately follow a normal (Gaussian) distribution.

1️⃣ Key Scenarios for Using Gaussian Naïve Bayes

✅ When features are continuous
GNB assumes that feature values are real numbers (e.g., height, weight, temperature).

Other variants (Multinomial/Bernoulli) work only with discrete features.

✅ When features are normally distributed (or close to it)

GNB works well if the data follows a bell-shaped curve.

If the data is highly skewed or has multiple peaks, GNB may not perform well.

✅ When you need a simple and fast probabilistic model

GNB is computationally efficient, even on large datasets.

Works well when interpretability is important.

✅ When you are dealing with medical or scientific datasets

Many biological and medical variables (e.g., blood pressure, cholesterol levels, heart rate) follow a normal distribution.

GNB is often used in disease prediction and diagnostic models.

✅ When you have small datasets

Naïve Bayes (including GNB) requires less training data than many other models.
It works well when there are few data points but a strong statistical relationship.

2️⃣ When NOT to Use Gaussian Naïve Bayes

❌ When features are categorical or discrete

Q24) Write a Python program to train an SVM Classifier with a Polynomial Kernel and visualize the decision
boundary:

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler

# Load the Iris dataset
iris = datasets.load_iris()
X = iris.data[:, :2]  # Use only the first two features for visualization
y = iris.target

# Select only two classes for simplicity (e.g., class 0 and class 1)
X = X[y != 2]
y = y[y != 2]

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Feature scaling (important for SVM with polynomial kernel)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Initialize the SVM classifier with Polynomial kernel
svm_poly = SVC(kernel='poly', degree=3, gamma='scale')

# Train the SVM model
svm_poly.fit(X_train, y_train)

# Define a function to plot the decision boundary
def plot_decision_boundary(X, y, model, title="Decision Boundary"):
    # Create a grid of points covering the feature space
    h = 0.02  # Step size in mesh
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1

    xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                         np.arange(y_min, y_max, h))

    # Predict labels for each point in the grid
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    # Create the plot
    plt.contourf(xx, yy, Z, alpha=0.8)
    plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors='k', marker='o', s=100, cmap=plt.cm.coolwarm)
    plt.title(title)
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.colorbar()

# Visualize the decision boundary
plot_decision_boundary(X_train, y_train, svm_poly, title="SVM Classifier with Polynomial Kernel")
plt.show()

# Evaluate the accuracy
accuracy = svm_poly.score(X_test, y_test)
print(f"Accuracy of SVM Classifier with Polynomial Kernel: {accuracy * 100:.2f}%")
```

Data splitting: We split the data into training and testing sets using train_test_split.

Feature Scaling: Feature scaling is applied using StandardScaler to standardize the features, which is important for SVM models, especially when using kernels like the polynomial kernel.

SVM Classifier: We initialize the SVM classifier with a polynomial kernel (kernel='poly') of degree 3 and train it on the training data.

Decision Boundary Visualization: The function plot_decision_boundary is used to visualize the decision boundary. We create a mesh grid covering the entire feature space and then predict the labels for each point in the grid. This helps to plot the decision boundary on the feature space.

Accuracy Evaluation: We calculate the accuracy of the classifier on the test data using score().

Output Example:

A plot will be generated showing the decision boundary created by the SVM classifier with the polynomial kernel. Points belonging to different classes will be shown in different colors, and the decision boundary will be the region that separates them.

The accuracy of the model will also be printed in the console.

The choice of polynomial degree (degree=3) is arbitrary here. You can experiment with different degrees to see how it affects the decision boundary and accuracy.

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# Step 1: Load the Iris dataset
data = load_iris()
X = data.data
y = data.target

# Step 2: Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Step 3: Train the Gaussian Naïve Bayes model without Laplace smoothing
gnb_no_smoothing = GaussianNB(var_smoothing=0)
gnb_no_smoothing.fit(X_train, y_train)

# Step 4: Predict with the model without smoothing
y_pred_no_smoothing = gnb_no_smoothing.predict(X_test)

# Step 5: Calculate accuracy of the model without smoothing
accuracy_no_smoothing = accuracy_score(y_test, y_pred_no_smoothing)
print(f"Accuracy without Laplace smoothing: {accuracy_no_smoothing * 100:.2f}%")

# Step 6: Train the Gaussian Naïve Bayes model with Laplace smoothing
# In Gaussian Naive Bayes, var_smoothing parameter is the equivalent of Laplace smoothing
# We will use a small value for the var_smoothing parameter to apply smoothing
gnb_with_smoothing = GaussianNB(var_smoothing=1e-9)
gnb_with_smoothing.fit(X_train, y_train)

# Step 7: Predict with the model after smoothing
y_pred_with_smoothing = gnb_with_smoothing.predict(X_test)

# Step 8: Calculate accuracy of the model after smoothing
accuracy_with_smoothing = accuracy_score(y_test, y_pred_with_smoothing)
print(f"Accuracy with Laplace smoothing: {accuracy_with_smoothing * 100:.2f}%")

# Step 9: Compare the predictions (before and after smoothing)
print("\nPredictions comparison (before vs after smoothing):")
for i in range(10):
    print(f"Sample {i+1}: Predicted (no smoothing) = {y_pred_no_smoothing[i]}, Predicted (with smoothing) = {y_pred_with_smoothing[i]}")
```

Q31) Write a Python program to train an SVM Classifier and use GridSearchCV to tune the hyperparameters (C,
gamma, kernel)

To train an SVM classifier and tune the hyperparameters (C, gamma, and kernel) using GridSearchCV, we'll use the GridSearchCV functionality from sklearn.model_selection. This allows us to search for the best combination of hyperparameters by exhaustively evaluating all possible combinations.

Here’s the step-by-step process for the program:

Load a dataset: We'll use the Iris dataset for simplicity, but you can use any classification dataset.

Define the SVM model: We'll use an SVC model (Support Vector Classifier) from sklearn.svm.

Define the parameter grid: We'll specify the hyperparameters (C, gamma, kernel) and the range of values to search over.

Use GridSearchCV to perform the exhaustive search for the best hyperparameters.

Train the model using the best hyperparameters.

Evaluate the model on the test set.

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Step 1: Load the Iris dataset
data = load_iris()
X = data.data
y = data.target

# Step 2: Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Step 3: Define the SVM model
svm = SVC()

# Step 4: Define the parameter grid to search over
param_grid = {
    'C': [0.1, 1, 10, 100],            # Regularization parameter
    'gamma': [0.001, 0.01, 0.1, 1],    # Kernel coefficient
    'kernel': ['linear', 'rbf', 'poly']  # Type of kernel
}

# Step 5: Set up GridSearchCV to find the best hyperparameters
grid_search = GridSearchCV(estimator=svm, param_grid=param_grid, cv=5, scoring='accuracy', n_jobs=-1)

# Step 6: Train the model with the best hyperparameters using GridSearchCV
grid_search.fit(X_train, y_train)

# Step 7: Output the best hyperparameters
print(f"Best hyperparameters: {grid_search.best_params_}")

# Step 8: Evaluate the best model on the test set
best_model = grid_search.best_estimator_
y_pred = best_model.predict(X_test)

# Step 9: Calculate and print the accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy of the best model: {accuracy * 100:.2f}%")
```

Train the SVM classifier using both One-vs-Rest (OvR) and One-vs-One (OvO) strategies.

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Step 1: Load the Wine dataset
data = load_wine()
X = data.data
y = data.target

# Step 2: Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Step 3: Train the SVM classifier using One-vs-Rest (OvR) strategy
svm_ovr = SVC(decision_function_shape='ovr', kernel='linear', random_state=42)
svm_ovr.fit(X_train, y_train)

# Step 4: Predict and evaluate accuracy for OvR
y_pred_ovr = svm_ovr.predict(X_test)
accuracy_ovr = accuracy_score(y_test, y_pred_ovr)

# Step 5: Train the SVM classifier using One-vs-One (OvO) strategy
svm_ovo = SVC(decision_function_shape='ovo', kernel='linear', random_state=42)
svm_ovo.fit(X_train, y_train)

# Step 6: Predict and evaluate accuracy for OvO
y_pred_ovo = svm_ovo.predict(X_test)
accuracy_ovo = accuracy_score(y_test, y_pred_ovo)

# Step 7: Compare the results
print(f"Accuracy with One-vs-Rest (OvR): {accuracy_ovr * 100:.2f}%")
print(f"Accuracy with One-vs-One (OvO): {accuracy_ovo * 100:.2f}%")
```

Q38) = Write a Python program to train an SVM Classifier using Stratified K-Fold Cross-Validation and compute the
average accuracy

To train an SVM classifier using Stratified K-Fold Cross-Validation and compute the average accuracy, we will use the Breast Cancer dataset from sklearn.datasets. Stratified K-Fold ensures that each fold has the same proportion of classes as the entire dataset, which is useful for imbalanced datasets.

Steps:

Load the Breast Cancer dataset.

Split the dataset using Stratified K-Fold Cross-Validation.

Train the SVM classifier on each fold.

Compute the average accuracy across all folds.

```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import StratifiedKFold
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Step 1: Load the Breast Cancer dataset
data = load_breast_cancer()
X = data.data
y = data.target

# Step 2: Initialize Stratified K-Fold Cross-Validation (5 folds)
kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Step 3: Initialize the SVM Classifier
svm_classifier = SVC(kernel='linear', random_state=42)

# Step 4: Perform cross-validation and compute accuracy for each fold
accuracies = []

for train_index, test_index in kf.split(X, y):
    # Split the data into training and test sets for this fold
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = y[train_index], y[test_index]

    # Train the SVM model
    svm_classifier.fit(X_train, y_train)

    # Predict on the test set
    y_pred = svm_classifier.predict(X_test)

    # Compute accuracy for this fold
    accuracy = accuracy_score(y_test, y_pred)
    accuracies.append(accuracy)

# Step 5: Compute and print the average accuracy across all folds
average_accuracy = np.mean(accuracies)
print(f"Average Accuracy with Stratified K-Fold Cross-Validation: {average_accuracy * 100:.2f}%")
```

Explanation of the Code:
Loading the Breast Cancer Dataset:

The Breast Cancer dataset is loaded using load_breast_cancer() from sklearn.datasets. This dataset is commonly used for binary classification (malignant vs benign).

Stratified K-Fold Cross-Validation:

We initialize a Stratified K-Fold object using StratifiedKFold(). It splits the data into 5 folds (n_splits=5), and ensures each fold has the same proportion of classes (malignant vs benign) as the entire dataset (shuffle=True and random_state=42 ensure reproducibility).

SVM Classifier:

We initialize an SVM classifier (SVC) with a linear kernel for simplicity (kernel='linear').

Training and Evaluation:

For each fold, the data is split into training and testing sets using the indices provided by StratifiedKFold. The SVM classifier is trained on the training data, and predictions are made on the test data.

We compute the accuracy for each fold using accuracy_score() and store the results in the accuracies list.

Average Accuracy:

After training and evaluating on all folds, we compute the average accuracy using np.mean() and print the result.

Load a dataset (e.g., the Iris dataset).

Split the dataset into training and testing sets.

Apply Recursive Feature Elimination (RFE) to select the best features based on model performance.

Train an SVM classifier on the original and reduced datasets.

Compare the accuracy of both models.

Load the Iris dataset.

Split the dataset into training and testing sets.

Train an SVM classifier without RFE to get the baseline accuracy.

Apply RFE to select the most important features.

Train an SVM classifier using the reduced set of features.

Compare the accuracy of both models.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.feature_selection import RFE
from sklearn.metrics import accuracy_score

# Step 1: Load the Iris dataset
data = load_iris()
X = data.data
y = data.target

# Step 2: Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Step 3: Train SVM Classifier without RFE (baseline model)
svm_classifier = SVC(kernel='linear', random_state=42)
svm_classifier.fit(X_train, y_train)

# Step 4: Predict and evaluate accuracy without RFE
y_pred_no_rfe = svm_classifier.predict(X_test)
accuracy_no_rfe = accuracy_score(y_test, y_pred_no_rfe)

# Step 5: Apply RFE (Recursive Feature Elimination) to select top 2 features
selector = RFE(svm_classifier, n_features_to_select=2)  # Select top 2 features
selector = selector.fit(X_train, y_train)

# Step 6: Get the reduced feature set
X_train_rfe = selector.transform(X_train)
X_test_rfe = selector.transform(X_test)

# Step 7: Train SVM Classifier on the reduced feature set
svm_classifier_rfe = SVC(kernel='linear', random_state=42)
svm_classifier_rfe.fit(X_train_rfe, y_train)

# Step 8: Predict and evaluate accuracy with RFE
y_pred_rfe = svm_classifier_rfe.predict(X_test_rfe)
accuracy_rfe = accuracy_score(y_test, y_pred_rfe)

# Step 9: Compare the results
print(f"Accuracy without RFE: {accuracy_no_rfe * 100:.2f}%")
print(f"Accuracy with RFE: {accuracy_rfe * 100:.2f}%")
```
